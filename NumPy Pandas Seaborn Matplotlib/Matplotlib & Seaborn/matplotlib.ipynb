```python

import matplotlib.pyplot as plt
import numpy as np

# Data for the pie chart
y = np.array([35, 25, 25, 15])

# Labels for each portion
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

# Beautiful mix of colors
mycolors = ["#FF6B6B", "#FFD93D", "#6BCB77", "#4D96FF"]

# Move the Apples portion outside the pie chart
myexplode = [0.15, 0, 0, 0]

# Create the pie chart
plt.pie(
    y,
    labels=mylabels,
    colors=mycolors,
    explode=myexplode,
    autopct='%1.1f%%',  # Display percentages
    startangle=90,
    shadow=True,
    wedgeprops={'edgecolor': 'black', 'linewidth': 1.5}
)

# Add title
plt.title("Fruit Distribution", fontsize=16, fontweight='bold')

# Display the pie chart
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Data for X-axis and Y-axis
x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])

# Set figure size
plt.figure(figsize=(9, 6))

# Create a red scatter plot
plt.scatter(
    x, y,
    color='red',          # Red dots
    s=120,                # Marker size
    marker='o',           # Circular markers
    edgecolors='black',   # Black border
    linewidths=1.5,      # Border thickness
    alpha=0.8,           # Slight transparency
    label='Data Points'
)

# Add title and axis labels
plt.title("Red Scatter Plot", fontsize=16, fontweight='bold')
plt.xlabel("X-Axis Values", fontsize=12)
plt.ylabel("Y-Axis Values", fontsize=12)

# Add grid lines
plt.grid(True, linestyle='--', alpha=0.4)

# Add legend
plt.legend()

# Automatically adjust layout
plt.tight_layout()

# Display the scatter plot
plt.show()
```

```python
# histogram

import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Generate 250 random values (mean=170, std=10)
np.random.seed(42)
x = np.random.normal(170, 10, 250)

# Set figure size
plt.figure(figsize=(10, 6))

# Create a histogram
plt.hist(
    x,
    bins=15,               # Number of bars
    color='red',           # Red bars
    edgecolor='black',     # Black borders
    linewidth=1.2,        # Border thickness
    alpha=0.8,            # Transparency
    label='Height Distribution'
)

# Calculate and display the mean
mean_value = np.mean(x)

plt.axvline(
    mean_value,
    color='blue',
    linestyle='dashed',
    linewidth=2,
    label=f'Mean: {mean_value:.2f} cm'
)

# Add title and axis labels
plt.title("Histogram of Student Heights",
          fontsize=16, fontweight='bold')

plt.xlabel("Height (cm)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

# Add grid lines
plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.4
)

# Display legend
plt.legend()

# Adjust layout
plt.tight_layout()

# Display histogram
plt.show()
```
