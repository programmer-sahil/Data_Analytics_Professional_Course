## 1. Import the required libraries

We use:

- **NumPy** → numerical work and creating sample data
- **Pandas** → tables and data analysis
- **Matplotlib** → visualization
- **Scikit-learn** → Machine Learning


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully.")
````


## 2. Create a simple retail dataset

For this beginner project, we create a **reproducible sample dataset** inside the notebook.

This means:
- no external file is required,
- the notebook runs directly in Google Colab,
- everyone gets the same data because we use `random_state` / a fixed random seed.

### Columns
| Column | Meaning |
|---|---|
| `Advertising_Spend` | Monthly advertising expense |
| `Website_Visitors` | Number of monthly website visitors |
| `Discount_Percent` | Average discount offered |
| `Previous_Month_Sales` | Sales from the previous month |
| `Monthly_Sales` | Value we want to predict |


````python
# Fixed seed = same random numbers every time
np.random.seed(42)

n = 200

advertising_spend = np.random.randint(10000, 100000, n)
website_visitors = np.random.randint(1000, 20000, n)
discount_percent = np.random.randint(0, 31, n)
previous_month_sales = np.random.randint(150000, 700000, n)

# Create realistic-looking monthly sales using a simple formula + random noise
noise = np.random.normal(0, 30000, n)

monthly_sales = (
    50000
    + 2.2 * advertising_spend
    + 8 * website_visitors
    + 1800 * discount_percent
    + 0.55 * previous_month_sales
    + noise
)

df = pd.DataFrame({
    "Advertising_Spend": advertising_spend,
    "Website_Visitors": website_visitors,
    "Discount_Percent": discount_percent,
    "Previous_Month_Sales": previous_month_sales,
    "Monthly_Sales": monthly_sales.round(2)
})

df.head()
````


## 3. Understand the dataset

Before Machine Learning, always inspect the data.

Important commands:

- `head()` → first rows
- `shape` → number of rows and columns
- `info()` → data types and missing values
- `describe()` → summary statistics


````python
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nSummary statistics:")
display(df.describe())
````


## 4. Exploratory Data Analysis (EDA)

EDA means **understanding the data before building the model**.

We will visualize how important features relate to sales.


````python
plt.figure(figsize=(7, 5))
plt.scatter(df["Advertising_Spend"], df["Monthly_Sales"], alpha=0.7)
plt.xlabel("Advertising Spend")
plt.ylabel("Monthly Sales")
plt.title("Advertising Spend vs Monthly Sales")
plt.grid(alpha=0.2)
plt.show()
````

````python
plt.figure(figsize=(7, 5))
plt.scatter(df["Previous_Month_Sales"], df["Monthly_Sales"], alpha=0.7)
plt.xlabel("Previous Month Sales")
plt.ylabel("Monthly Sales")
plt.title("Previous Month Sales vs Current Monthly Sales")
plt.grid(alpha=0.2)
plt.show()
````


## 5. Select Features (`X`) and Target (`y`)

### Feature
An input used by the model.

### Target
The value the model tries to predict.

Here:

- `X` = Advertising Spend, Website Visitors, Discount %, Previous Month Sales
- `y` = Monthly Sales


````python
X = df[
    [
        "Advertising_Spend",
        "Website_Visitors",
        "Discount_Percent",
        "Previous_Month_Sales"
    ]
]

y = df["Monthly_Sales"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
````


## 6. Split the data into Training and Testing sets

We should **not train and test on the same data**.

A common beginner split is:

- **80% Training Data** → model learns from this
- **20% Testing Data** → model is evaluated on unseen data

`random_state=42` makes the split reproducible.


````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
````


## 7. Create and train the Linear Regression model

### What is Linear Regression?

Linear Regression predicts a **continuous numerical value**.

Examples:
- Sales
- Salary
- House price
- Revenue
- Demand

The most important Scikit-learn pattern is:

```python
model = Algorithm()
model.fit(X_train, y_train)
prediction = model.predict(X_test)
```

### Easy memory rule
**Create → Fit → Predict → Evaluate**


````python
model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("Model training completed.")
````


## 8. Make predictions

The model has learned patterns from the training data.

Now we ask it to predict sales for the testing data.


````python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual_Sales": y_test.values,
    "Predicted_Sales": y_pred
})

results.head(10)
````


## 9. Evaluate the model

For regression, remember these three metrics:

### MAE — Mean Absolute Error
Average prediction error.

**Lower is better.**

### RMSE — Root Mean Squared Error
Similar to MAE, but larger errors get more penalty.

**Lower is better.**

### R² Score
Shows how much variation in the target is explained by the model.

- closer to **1** → stronger fit
- around **0** → weak explanatory power
- can be negative → model performs poorly


````python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE : {mae:,.2f}")
print(f"RMSE: {rmse:,.2f}")
print(f"R²  : {r2:.4f}")
````


## 10. Visualize Actual vs Predicted Sales

A good prediction model should produce points reasonably close to the diagonal reference line.


````python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, alpha=0.75)

minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot([minimum, maximum], [minimum, maximum], linestyle="--")

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Actual vs Predicted Sales")
plt.grid(alpha=0.2)
plt.show()
````


## 11. Understand the model coefficients

Linear Regression gives each feature a **coefficient**.

A larger positive coefficient means that, **holding other model inputs constant**, increasing that feature is associated with a larger predicted sales value.

> Do not automatically interpret coefficients as proof of causation.


````python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
}).sort_values("Coefficient", ascending=False)

feature_importance
````


## 12. Predict sales for a new business scenario

Suppose a store has:

- Advertising Spend = ₹60,000
- Website Visitors = 12,000
- Discount = 15%
- Previous Month Sales = ₹4,50,000

We can use the trained model to estimate the next month's sales.


````python
new_store = pd.DataFrame({
    "Advertising_Spend": [60000],
    "Website_Visitors": [12000],
    "Discount_Percent": [15],
    "Previous_Month_Sales": [450000]
})

predicted_sales = model.predict(new_store)[0]

print(f"Predicted Monthly Sales: ₹{predicted_sales:,.2f}")
````


# ✅ Complete Machine Learning Workflow

Remember this sequence:

**Problem → Data → Cleaning → EDA → X & y → Train/Test Split → Model → Fit → Predict → Evaluate → Improve**

For Scikit-learn, memorize:

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

```



# 🎯 Interview Questions — Quick Revision

### 1. What is Machine Learning?
Machine Learning allows computers to learn patterns from data and make predictions or decisions.

### 2. What is supervised learning?
A type of ML where training data contains input features and a known target/output.

### 3. Why did you use Linear Regression?
Because the target, **Monthly Sales**, is a continuous numerical value.

### 4. What are `X` and `y`?
- `X` = input features
- `y` = target to predict

### 5. Why do we split the dataset?
To train the model on one part and test its performance on unseen data.

### 6. What does `fit()` do?
It trains the model using the training data.

### 7. What does `predict()` do?
It uses the trained model to generate predictions for new input data.

### 8. What is MAE?
The average absolute difference between actual and predicted values.

### 9. What is R²?
A regression metric that measures how much variance in the target is explained by the model.

### 10. What is overfitting?
When a model performs very well on training data but poorly on unseen data.
